# 03 - Missing Value Handling

Still working with `hr_employee_attrition_raw.csv`. Five columns have real gaps in them - `MonthlyIncome`, `JobSatisfaction`, `Education`, `DistanceFromHomeKM`, `Gender` — and my goal here isn't to just plug every hole the same way. I want to pick a method per column based on what's actually going on with it.

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
df = pd.read_csv('hr_employee_attrition_raw.csv')
df.shape
```

## 1. What Are Missing Values?

A missing value is just an absence where a value should logically be there. Pandas represents this as `NaN` for numeric/object columns, `NaT` for datetime, or `None`. It's worth separating this from a value like `0` or `"Unknown"` — those are actual data points someone recorded. Missing means nothing was recorded at all.

```python
df.isna().sum().loc[lambda s: s > 0]
```

## 2. Why Missing Values Occur

Usually it's something mundane: a form field someone skipped, a system export that dropped a field, a survey question left blank, a typo during manual entry, or someone deliberately not answering. What matters more than the fact that something's missing is *why* — that's what tells you whether imputing it is even a reasonable thing to do, and with what method. There are three standard categories:

### MCAR — Missing Completely At Random
The chance of a value being missing has **no relationship** to any variable, observed or unobserved. Example: a survey response lost due to a random system glitch. Since the missing subset is a random sample of the whole, dropping or simple-imputing it introduces little bias.

### MAR — Missing At Random
The chance of missingness depends on **other observed variables**, but not on the missing value itself. Example: newer employees (a variable we *do* have, `YearsAtCompany`) may be less likely to have completed a `JobSatisfaction` survey yet — missingness relates to tenure, not to the satisfaction score itself.

### MNAR — Missing Not At Random
The chance of missingness depends on the **value that is missing**, even after accounting for other variables. Example: employees with very low `JobSatisfaction` may be the ones who skip the satisfaction survey altogether — the missingness is tied to the unobserved answer itself. This is the hardest case: imputation can systematically underestimate dissatisfaction no matter which method is used.

Classifying missingness is rarely provable from data alone — it requires domain reasoning about *how* the data was collected.

```python
# A simple MAR-style check: does missingness in JobSatisfaction relate to an observed variable (YearsAtCompany)?
df.assign(js_missing=df['JobSatisfaction'].isna()).groupby('js_missing')['YearsAtCompany'].apply(lambda s: pd.to_numeric(s.astype(str).str.replace(' yrs','',regex=False), errors='coerce').mean())
```

If the average tenure looks pretty different between the missing and non-missing groups, that's a sign missingness is tangled up with an observed variable (leaning MAR) rather than being pure random noise (MCAR).

## 3. Detecting Missing Values

```python
df.isna().sum()
```

```python
df.isna().any(axis=1).sum(), df.isna().sum(axis=1).value_counts()
```

`isna().sum(axis=1)` is one I use a lot — it counts how many gaps each *row* has, which is a fast way to spot rows that are missing several things at once. Those tend to be lower-quality records overall, not just unlucky on one field.

## 4. Missing Value Percentage

Raw counts don't mean much on their own — turning it into a percentage of total rows makes it possible to actually compare severity across columns.

```python
missing_pct = (df.isna().sum() / len(df) * 100).round(2)
missing_pct[missing_pct > 0].sort_values(ascending=False)
```

Rough guide I go by (not a hard rule):
- **Under 5%** — dropping or imputing rarely makes a real difference either way
- **5-30%** — I'd lean toward imputing over dropping, but the method needs thought
- **Over 40-50%** — worth asking if the column is even worth keeping

Everything here is under 5%, so technically I *could* just drop rows for all five columns. But "technically allowed" isn't the same as "the right call" — that's the whole point of walking through the individual methods below instead of defaulting to one.

## 5. Techniques for Handling Missing Values

Going through each one with when to use it, when not to, what it's good at, and where it falls short — and applying each to whichever column actually fits it best.

### 5.1 Dropping Rows (`dropna()`)

**When to use:** Missingness is MCAR, the affected proportion is small (roughly <5%), and the dataset is large enough that losing those rows doesn't hurt statistical power.

**When not to use:** Missingness is MAR/MNAR (dropping biases the remaining sample), the proportion is large, or multiple columns are missing on overlapping rows (compounding row loss).

**Advantages:** Simple, introduces no artificial values, keeps analysis honest about what was actually observed.

**Limitations:** Reduces sample size, throws away potentially useful information in the row's other columns, and can bias results if missingness isn't MCAR.

```python
before = len(df)
df_dropped_rows = df.dropna(subset=['Gender'])
print(f'{before} -> {len(df_dropped_rows)} rows ({before - len(df_dropped_rows)} dropped)')
```

### 5.2 Dropping Columns

**When to use:** A column is missing so much (typically over 40-50%) that imputing it would mostly be guesswork, or the column just isn't important enough to bother fixing.

**When not to use:** The column matters even with the gaps, or the missing percentage isn't that bad.

**Advantages:** Gets rid of a weak/unreliable feature cleanly.

**Limitations:** Whatever real signal was in there is gone for good — a bad call here can't really be undone later.

Nothing in this dataset qualifies — `MonthlyIncome` is the worst offender and it's still under 5%. The check below is really just the reusable pattern for *finding* columns that would qualify, since none do here.

```python
threshold = 40
cols_to_drop = missing_pct[missing_pct > threshold].index.tolist()
print(f'Columns exceeding {threshold}% missing:', cols_to_drop if cols_to_drop else 'none')
```

### 5.3 Mean Imputation

**When to use:** Numeric column, missingness is roughly MCAR, and the distribution is fairly symmetric without extreme outliers.

**When not to use:** The distribution is skewed or has outliers (mean gets pulled away from the typical value), or the column is categorical.

**Advantages:** Simple, preserves the overall mean of the column, fast to compute.

**Limitations:** Shrinks variance (all imputed rows get an identical value), can distort the mean's relationship with other variables, and is a poor fit for skewed data.

```python
income_numeric = pd.to_numeric(df['MonthlyIncome'].astype(str).str.replace(' USD', '', regex=False), errors='coerce')
print(income_numeric.describe())
income_mean_imputed = income_numeric.fillna(income_numeric.mean())
print('Missing before:', income_numeric.isna().sum(), '| after:', income_mean_imputed.isna().sum())
```

Once the malformed entries are filtered out, `MonthlyIncome` is close enough to symmetric that mean imputation is a reasonable call. That's the actual justification — it's not something I'd reach for automatically.

### 5.4 Median Imputation

**When to use:** Numeric column that is skewed or contains outliers — the median is robust to extreme values that would drag a mean off-center.

**When not to use:** The column is not skewed (mean would work just as well and uses more information), or the column is categorical.

**Advantages:** Robust to outliers, better represents the "typical" value in skewed data than the mean does.

**Limitations:** Still collapses variance by assigning one repeated value, and ignores relationships with other columns.

```python
print(df['DistanceFromHomeKM'].describe())
distance_median_imputed = df['DistanceFromHomeKM'].fillna(df['DistanceFromHomeKM'].median())
print('Missing before:', df['DistanceFromHomeKM'].isna().sum(), '| after:', distance_median_imputed.isna().sum())
```

`DistanceFromHomeKM` maxes out around 830 with a median sitting near 5 — that's a serious right skew (same outliers I flagged back in Notebook 1). Median is the safer pick here; the mean would get dragged toward those extreme values.

### 5.5 Mode Imputation

**When to use:** Categorical column with a clearly dominant category and MCAR-style missingness.

**When not to use:** The column has many similarly-frequent categories (mode is nearly arbitrary), or missingness itself carries meaning (then a constant/'Unknown' label is more honest — see 5.6).

**Advantages:** Simple, keeps the column categorical, doesn't invent a new category.

**Limitations:** Over-represents the already-most-common category, distorting the category distribution; ignores every other column when choosing the fill value.

```python
print(df['Education'].value_counts(dropna=False))
education_mode = df['Education'].mode()[0]
education_mode_imputed = df['Education'].fillna(education_mode)
print('Mode used:', education_mode)
print('Missing before:', df['Education'].isna().sum(), '| after:', education_mode_imputed.isna().sum())
```

### 5.6 Constant Value Imputation

**When to use:** Missingness is likely meaningful rather than random (potential MNAR), and it's more honest to explicitly label it than to guess a value. Common for categorical columns (`'Unknown'`, `'Not Specified'`) or as a paired missing-value flag for numeric columns.

**When not to use:** As a numeric filler when the constant would visibly distort statistics (e.g. filling missing income with `0` wrecks the mean and any model using it).

**Advantages:** Transparent — doesn't pretend to know a value it doesn't. Preserves the fact that data was missing, which a model or analyst can use as its own signal.

**Limitations:** Adds a new category that has no real-world meaning; for numeric columns, an arbitrary constant like `-1` must be handled carefully so models don't treat it as an ordinary number.

```python
gender_constant_imputed = df['Gender'].fillna('Not Specified')
gender_constant_imputed.value_counts(dropna=False)
```

For `Gender`, I'd guess the missing values are more likely someone declining to answer than a random glitch. Labeling it `'Not Specified'` is more honest than guessing Male or Female off the mode.

### 5.7 Forward Fill (`ffill`)

**When to use:** Data has a meaningful order (typically time series or sequential logs) where a missing value is likely the same as the last known value — e.g. a sensor reading that stays constant between measurements.

**When not to use:** Rows aren't in a meaningful sequence, or there's no reason to believe consecutive rows share a value. Applying it to arbitrarily-ordered tabular data (like this employee list, which has no natural row order) fills with a neighbor's value for no real reason.

**Advantages:** Fast, works well for genuinely sequential/time-ordered data (stock prices, sensor logs).

**Limitations:** Propagates errors forward if the last known value was wrong; meaningless on unordered data; can't fill a leading `NaN` with nothing before it.

```python
# Correct use: sort by JoinDate first so row order is actually meaningful, then demonstrate on a copy
demo = df[['EmployeeID', 'JoinDate', 'JobSatisfaction']].copy()
demo['JoinDate'] = pd.to_datetime(demo['JoinDate'])
demo = demo.sort_values('JoinDate').reset_index(drop=True)
demo['JobSatisfaction_ffill'] = demo['JobSatisfaction'].ffill()
demo[demo['JobSatisfaction'].isna()].head()
```

Sorting by `JoinDate` doesn't actually fix anything here — `JobSatisfaction` isn't the kind of thing that carries over from 'whoever got hired right before this person.' I'm including this on purpose as an example of forward fill being used where it doesn't belong. It runs without error, which is exactly the trap: the code working doesn't mean the logic is sound.

### 5.8 Backward Fill (`bfill`)

**When to use:** Same conditions as forward fill, but when the *next* value is the more reasonable stand-in — e.g. filling a reading with the next known reading when a process starts partway through.

**When not to use:** Same caveats as forward fill — requires genuine sequential order and a real reason to expect adjacent-row similarity.

**Advantages:** Simple, useful complement to `ffill` for filling leading gaps `ffill` can't reach.

**Limitations:** Uses "future" information relative to the row's position — for genuine time series this can leak information that wouldn't have been available yet.

```python
demo['JobSatisfaction_bfill'] = demo['JobSatisfaction'].bfill()
demo[['EmployeeID','JoinDate','JobSatisfaction','JobSatisfaction_ffill','JobSatisfaction_bfill']].head(8)
```

### 5.9 Interpolation

**When to use:** Numeric data with a genuine order (time series, measurements along a continuum) where values change gradually, so a value between two known points can reasonably be estimated by a line (or curve) between them.

**When not to use:** Categorical data, unordered data, or numeric data with abrupt jumps rather than smooth trends.

**Advantages:** Produces smoother, more plausible fill values than a single constant for trending numeric data; several methods available (`linear`, `time`, `polynomial`).

**Limitations:** Requires a meaningful order to be valid; assumes smooth/linear change between points, which may not hold; can extrapolate badly near the edges of the data.

```python
# Demonstrate on a genuinely time-ordered numeric series: average income by join-month
by_month = (df.assign(JoinDate=pd.to_datetime(df['JoinDate']),
                       Income=pd.to_numeric(df['MonthlyIncome'].astype(str).str.replace(' USD','',regex=False), errors='coerce'))
              .set_index('JoinDate')['Income']
              .resample('ME').mean())
by_month_interp = by_month.interpolate(method='linear')
pd.DataFrame({'avg_income': by_month, 'interpolated': by_month_interp}).head(10)
```

Calendar months are a real, meaningful order, so estimating a missing month as somewhere between its neighbors actually holds up — unlike the JobSatisfaction example above.

### 5.10 Group-Based Imputation

**When to use:** A meaningful grouping variable exists and the missing column's typical value differs across groups — e.g. income varies by department, so a department-level median is more accurate than one global median.

**When not to use:** No relevant grouping variable exists, or group sizes are too small to give a stable statistic (a group of 2 employees gives an unreliable median).

**Advantages:** More accurate than a single global fill value because it uses other observed information (MAR-aware); still simple to compute and interpret.

**Limitations:** Requires choosing a sensible group; small groups produce unstable estimates; still assigns identical values within a group, understating within-group variability.

```python
dept_median_income = df.groupby('Department')['MonthlyIncome'].apply(
    lambda s: pd.to_numeric(s.astype(str).str.replace(' USD','',regex=False), errors='coerce').median()
)
dept_median_income
```

```python
income_clean = pd.to_numeric(df['MonthlyIncome'].astype(str).str.replace(' USD','',regex=False), errors='coerce')
income_group_imputed = income_clean.fillna(df['Department'].map(dept_median_income))
print('Missing before:', income_clean.isna().sum(), '| after group-based fill:', income_group_imputed.isna().sum())
```

### 5.11 KNN Imputation

**When to use:** Numeric columns where similar rows (by other features) likely have similar values, and the dataset isn't too large for pairwise distance computation to be expensive.

**When not to use:** Very large datasets (KNN distance computation scales poorly), mostly categorical data without numeric encoding, or when features used for distance aren't actually informative about the missing one.

**Advantages:** Uses multivariate information rather than a single global or group statistic, often more accurate when features are correlated.

**Limitations:** Computationally expensive on large data, sensitive to feature scaling (features should typically be standardized first), and the choice of `k` and distance metric affects results.

```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler

num_cols = ['Age', 'JobSatisfaction', 'PerformanceRating', 'DistanceFromHomeKM']
num_df = df[num_cols].copy()

scaler = StandardScaler()
scaled = scaler.fit_transform(num_df)

knn_imputer = KNNImputer(n_neighbors=5)
imputed_scaled = knn_imputer.fit_transform(scaled)
imputed = pd.DataFrame(scaler.inverse_transform(imputed_scaled), columns=num_cols)

print('JobSatisfaction missing before:', num_df['JobSatisfaction'].isna().sum(), '| after KNN:', imputed['JobSatisfaction'].isna().sum())
imputed.loc[num_df['JobSatisfaction'].isna(), 'JobSatisfaction'].describe()
```

Scaling matters here because KNN measures distance across all four columns at once. Without it, `DistanceFromHomeKM` (values running into the hundreds) would swamp `JobSatisfaction` (a 1-5 scale) in the distance calculation, and the neighbors it picks would barely reflect satisfaction at all.

### 5.12 Iterative Imputation

**When to use:** Numeric columns with meaningful multivariate relationships, where modeling each missing column as a function of the others captures more signal than KNN or a group statistic — good for moderately-sized datasets with several correlated numeric columns.

**When not to use:** Very few numeric columns to model against (little to learn from), very large datasets (iterative model fitting is slower than simpler methods), or when interpretability of the fill logic matters more than accuracy.

**Advantages:** Often the most accurate of the numeric methods, since it explicitly models each column's relationship with the others, iterating until estimates stabilize.

**Limitations:** Computationally heaviest option here, results can vary with the underlying regression model and random state, and it's the hardest method to explain to a non-technical audience.

```python
from sklearn.experimental import enable_iterative_imputer  # noqa: F401
from sklearn.impute import IterativeImputer

iter_imputer = IterativeImputer(random_state=42, max_iter=10)
iter_result = pd.DataFrame(iter_imputer.fit_transform(num_df), columns=num_cols)

print('JobSatisfaction missing before:', num_df['JobSatisfaction'].isna().sum(), '| after iterative:', iter_result['JobSatisfaction'].isna().sum())
iter_result.loc[num_df['JobSatisfaction'].isna(), 'JobSatisfaction'].describe()
```

## 6. Comparing the Fills

None of these methods are really swappable for one another. Lining a few up side by side on `JobSatisfaction` shows just how much the choice moves the actual imputed numbers:

```python
comparison = pd.DataFrame({
    'mean_fill': num_df['JobSatisfaction'].fillna(num_df['JobSatisfaction'].mean()),
    'median_fill': num_df['JobSatisfaction'].fillna(num_df['JobSatisfaction'].median()),
    'knn_fill': imputed['JobSatisfaction'],
    'iterative_fill': iter_result['JobSatisfaction'],
})[num_df['JobSatisfaction'].isna()]
comparison.describe()
```
