# 02 - Data Type Handling

Same file as before: `hr_employee_attrition_raw.csv`. I'm reusing it on purpose — it still has all the string-encoded numbers and unit suffixes I ran into in Notebook 1, and those are exactly what makes this topic worth practicing on real data instead of a toy example.

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
df = pd.read_csv('hr_employee_attrition_raw.csv')
print(df.shape)
df.head(3)
```

## 1. Identifying Data Types

First step is always `df.dtypes`. It tells you what pandas *thinks* each column is right now — which isn't necessarily what it should be.

```python
df.dtypes
```

`MonthlyIncome` and `YearsAtCompany` show up as `object`/string even though they're clearly numbers, and `JoinDate` is a string even though it's a date. I see this constantly with data that's come out of a spreadsheet or a form export — the values look fine to a human, pandas just doesn't know any better yet.

```python
df.select_dtypes(include='number').columns.tolist(), df.select_dtypes(include=['object', 'str']).columns.tolist()
```

## 2. Numerical Data

Numerical columns are the ones you can do math on — mean, sum, comparisons, scaling. There are really two flavors here:
- **Continuous** — `DistanceFromHomeKM`, `MonthlyIncome` (once it's fixed)
- **Discrete / counts** — `Age`, `YearsAtCompany`, `PerformanceRating`

The catch: pandas only treats a column as numeric if the dtype is actually `int64` or `float64`. A column full of numeric-looking text doesn't count, no matter how it looks when you print it.

```python
df['Age'].dtype, df['DistanceFromHomeKM'].dtype, df['MonthlyIncome'].dtype
```

```python
try:
    df['MonthlyIncome'].mean()
except TypeError as e:
    print('Error:', e)
```

And there's the proof — `MonthlyIncome` reads like numbers but it's text under the hood, so `.mean()` just refuses. Good reminder to check `dtype` instead of trusting what the column looks like on screen.

## 3. Categorical Data

Categorical data means a fixed, small set of labels — think `Department`, `JobRole`, `Education`, `Gender`. Instead of leaving these as generic `object` strings, pandas has a dedicated `category` dtype, and switching to it buys a few things:
- Less memory (under the hood it's just storing integer codes)
- Makes the intent obvious to anyone reading the code, or to plotting libraries
- Lets you do ordered comparisons when the categories actually have an order (education level being the obvious one here)

```python
df['Department'].nunique(), df['Department'].unique()
```

```python
df['Department'] = df['Department'].astype('category')
df['Department'].dtype
```

Education levels have a real order to them, so instead of a plain category I can tell pandas the order explicitly:

```python
education_order = ['High School', "Bachelor's", "Master's", 'PhD']
df['Education'] = pd.Categorical(df['Education'], categories=education_order, ordered=True)
df['Education'].dtype
```

```python
# Ordered categories support comparison operators
df[df['Education'] > 'High School']['Education'].value_counts()
```

## 4. Boolean Data

A boolean column only has two real states. `OverTime` and `Attrition` are conceptually booleans, but they show up as messy Yes/No strings with different casing.

```python
df['OverTime'].unique()
```

```python
overtime_clean = df['OverTime'].str.strip().str.lower().map({'yes': True, 'y': True, 'no': False, 'n': False})
overtime_clean.dtype, overtime_clean.value_counts(dropna=False)
```

Turning that into actual `True`/`False` values — instead of leaving six different spellings of 'yes' floating around — means I can just write `df[overtime_clean]` later instead of remembering every variant.

## 5. Date/Time Data

`JoinDate` arrives as plain text (`'2019-07-23'`). As a string it's basically frozen — no computing tenure, no sorting by how recent someone joined, no pulling out the year.

```python
df['JoinDate'].dtype
```

```python
join_date = pd.to_datetime(df['JoinDate'])
join_date.dtype
```

```python
# Now date arithmetic works
tenure_days = (pd.Timestamp('2025-01-01') - join_date).dt.days
tenure_days.describe()
```

```python
join_date.dt.year.value_counts().sort_index()
```

## 6. String Data

`EmployeeID`, `JobRole`, and anything free-text-ish stays as a string (`object` in pandas terms). That's the right call for identifiers and labels — you're never going to average an `EmployeeID`. The `.str` accessor gives vectorized string operations once it's in this form.

```python
df['EmployeeID'].dtype
```

```python
df['EmployeeID'].str.startswith('EMP').all(), df['EmployeeID'].str.len().unique()
```

## 7. Converting Data Types

### `astype()`
Good for values that are already clean — an int-to-float cast, or object-to-category. It has zero tolerance for messy input though: one bad value and the whole thing throws an error.

```python
# Integer -> Float
df['PerformanceRating'].astype(float).dtype
```

```python
# astype() fails hard on invalid values - it does not silently coerce
try:
    df['MonthlyIncome'].astype(float)
except ValueError as e:
    print('astype() failed:', str(e)[:120])
```

### `to_numeric()`
This is the one built specifically for numeric conversion, and it has an `errors=` argument that controls what happens when something doesn't convert:
- `errors='raise'` (the default) — stop at the first bad value
- `errors='coerce'` — turn anything unconvertible into `NaN`
- `errors='ignore'` — leave the column as-is (this is on its way out in newer pandas, better to just try/except)

```python
# String -> Numeric, after stripping unit suffixes
income_clean = df['MonthlyIncome'].str.replace(' USD', '', regex=False)
income_numeric = pd.to_numeric(income_clean, errors='coerce')
print(income_numeric.dtype)
print('New NaNs introduced by bad values:', income_numeric.isna().sum() - df['MonthlyIncome'].isna().sum())
```

```python
# Inspect what actually failed to convert
still_bad = df.loc[income_numeric.isna() & df['MonthlyIncome'].notna(), 'MonthlyIncome']
still_bad.head(10)
```

### `to_datetime()`
Same idea as `to_numeric()`, but for dates — same `errors=` control, plus a `format=` argument so you can tell it exactly what pattern to expect instead of making it guess.

```python
# String -> Date
join_date_coerced = pd.to_datetime(df['JoinDate'], format='%Y-%m-%d', errors='coerce')
join_date_coerced.isna().sum()
```

### Numeric -> Categorical
Bucketing a continuous number into labeled bands (`pd.cut`) is technically a conversion too — turns a raw number into a category. Handy when a report or model needs groups instead of exact values.

```python
# Numeric -> Categorical
age_band = pd.cut(df['Age'].clip(upper=75), bins=[0, 25, 35, 45, 100], labels=['<=25', '26-35', '36-45', '46+'])
age_band.dtype, age_band.value_counts()
```

## 8. Handling Invalid Conversions

In practice, columns almost never convert cleanly on the first attempt. The pattern I keep coming back to:
1. Convert with `errors='coerce'`
2. Compare the new `NaN` count against how many were already missing *before* converting
3. Actually go look at what caused the difference before deciding what to do about it

```python
def coerce_report(original: pd.Series, converted: pd.Series, name: str):
    new_nans = converted.isna().sum() - original.isna().sum()
    print(f'{name}: {new_nans} value(s) failed to convert and became NaN')
    if new_nans > 0:
        print(original[converted.isna() & original.notna()].unique()[:10])

coerce_report(df['MonthlyIncome'], income_numeric, 'MonthlyIncome')
```

**The one thing I try not to do is coerce and just move on.** Quietly turning bad values into `NaN` can bury a real data entry mistake (that `'41019.93508181599'` income value looks like a plausible float, but it's suspicious) or a formatting quirk (`'5208.92 USD'`) that deserves its own fix rather than getting lumped in with genuinely missing data.

## 9. Detecting Incorrect Data Types

Three checks I find useful for catching columns that are the wrong dtype without having to eyeball every one manually:

```python
# Check 1: object columns that are actually fully numeric
for col in df.select_dtypes(include=['object', 'str']).columns:
    converted = pd.to_numeric(df[col], errors='coerce')
    pct_numeric = converted.notna().mean() * 100
    if pct_numeric > 80:
        print(f"{col}: {pct_numeric:.1f}% of values parse as numeric but dtype is {df[col].dtype}")
```

```python
# Check 2: low-cardinality object columns that should be categorical
for col in df.select_dtypes(include=['object', 'str']).columns:
    n_unique = df[col].nunique()
    if n_unique < 15:
        print(f'{col}: only {n_unique} unique values -> candidate for category dtype')
```

```python
# Check 3: object columns that actually hold dates
for col in df.select_dtypes(include=['object', 'str']).columns:
    parsed = pd.to_datetime(df[col], format='%Y-%m-%d', errors='coerce')
    if parsed.notna().mean() > 0.9:
        print(f'{col}: {parsed.notna().mean()*100:.1f}% parse as dates -> candidate for datetime64')
```

## 10. Why Correct Data Types Matter

- **Computations actually work** — `.mean()`, `.sum()`, comparisons — all of it either breaks or silently gives the wrong answer on numeric-looking strings (string sorting, not numeric sorting).
- **Memory** — `category` dtype and right-sized numeric types (`int32` instead of `int64` when you don't need the range) add up on bigger datasets.
- **Models expect it** — ML algorithms want numeric input. Dates and booleans need encoding, and that step assumes the column is already the right semantic type going in.
- **Sorting and grouping behave correctly** — a numeric column stored as text sorts `'10' < '2'` alphabetically, which quietly wrecks any ranking or groupby that relies on numeric order.
- **It surfaces problems early** — honestly, half the time I find messy values (stray units, wrong formats) is by *trying* to convert a column and watching what breaks, not by inspecting it first.
