Dataset used: `hr_employee_attrition_raw.csv`. Contains numerical features, categorical features, missing values, duplicates, outliers, mixed data types, inconsistent category values, and irrelevant columns.




## 1. Data Quality Issues in This Dataset

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
df_raw = pd.read_csv('hr_employee_attrition_raw.csv')
print(df_raw.shape)
df_raw.head()
```

```python
df_raw.dtypes
```

**Issue 1: Wrong data types.** `MonthlyIncome` and `YearsAtCompany` are text instead of numeric due to unit suffixes.

```python
df_raw[df_raw['MonthlyIncome'].astype(str).str.contains('USD', na=False)][['EmployeeID','MonthlyIncome']].head()
```

```python
df_raw[df_raw['YearsAtCompany'].astype(str).str.contains('yrs', na=False)][['EmployeeID','YearsAtCompany']].head()
```

**Issue 2: Missing values.**

```python
missing = df_raw.isna().sum()
missing_pct = (missing / len(df_raw) * 100).round(2)
pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})[missing > 0].sort_values('missing_count', ascending=False)
```

**Issue 3: Inconsistent category spellings.**

```python
print(sorted(df_raw['Gender'].dropna().unique()))
print(sorted(df_raw['Department'].dropna().unique()))
print(sorted(df_raw['OverTime'].dropna().unique()))
```

**Issue 4: Outliers and impossible values.** `Age` has values like 999 and -5. `MonthlyIncome` and `DistanceFromHomeKM` have extreme values.

```python
print(df_raw['Age'].describe())
sorted(df_raw.loc[(df_raw['Age'] > 90) | (df_raw['Age'] < 16), 'Age'].unique())
```

```python
df_raw['DistanceFromHomeKM'].describe()
```

**Issue 5: Duplicate records.**

```python
exact_dupes = df_raw.duplicated().sum()
business_dupes = df_raw.drop(columns=['EmployeeID']).duplicated().sum()
print(exact_dupes, business_dupes)
```

**Issue 6: Irrelevant features.** `EmployeeCountFlag` is constant. `RandomSurveyCode` is a meaningless ID.

```python
print(df_raw['EmployeeCountFlag'].unique())
df_raw['RandomSurveyCode'].head()
```

## 2. What is Data Preprocessing

Data preprocessing converts raw data into a clean, consistent format suitable for analysis or modeling. It fixes data types, handles missing values, standardizes categories, removes duplicates, treats outliers, and drops irrelevant columns.


## 3. Why Data Preprocessing is Required

1. Wrong patterns get learned from bad data.
2. Algorithms require specific data types.
3. Missing values break computation.
4. Duplicates distort statistics.
5. Outliers skew averages and distance-based models.
6. Irrelevant features add noise and reduce model performance.


```python
try:
    df_raw['MonthlyIncome'].mean()
except TypeError as e:
    print(e)
```

## 4. Raw Data vs Clean Data

**Data types:** Raw data has text mixed with numbers. Clean data has proper numeric types.

**Missing values:** Raw data has missing values unhandled. Clean data has them imputed or removed.

**Categories:** Raw data has multiple spellings for the same category. Clean data has one standard spelling.

**Duplicates:** Raw data has duplicate records present. Clean data has duplicates removed.

**Outliers:** Raw data has outliers unchecked. Clean data has outliers validated against a realistic range.

**Columns:** Raw data includes irrelevant columns. Clean data includes only informative columns.

## 5. Data Quality Pillars

**Consistency:** Same fact represented the same way everywhere. Violated by Gender, Department, OverTime spelling variants.

**Completeness:** All required values present. Violated by nulls in MonthlyIncome, JobSatisfaction, etc.

**Accuracy:** Values reflect reality. Violated by Age = 999.

**Validity:** Values match expected type or range. Violated by MonthlyIncome = "5208.92 USD".

**Integrity:** Relationships and constraints hold across the dataset. Violated by duplicate records with different EmployeeID values.


```python
completeness = (1 - df_raw.isna().mean()) * 100
completeness.round(2)
```

```python
invalid_age_mask = ~df_raw['Age'].between(16, 75)
print(invalid_age_mask.sum())
df_raw.loc[invalid_age_mask, ['EmployeeID', 'Age']]
```

## 6. Data Leakage

Data leakage occurs when information unavailable at prediction time enters the training process, giving false confidence in model performance.

**Target leakage:** A feature derived from the target itself. Example: including "ExitInterviewCompleted" when predicting Attrition.

**Train/test contamination:** Preprocessing statistics computed on the full dataset instead of training data only.

**Temporal leakage:** Using future information to predict the past.

**Rule:** Fit all preprocessing statistics on the training split only, then apply to validation and test splits.


## 7. Training, Validation, and Test Data

| Split | Purpose | Typical Size |
|-------|---------|---------------|
| Training | Fit the model | 60-70% |
| Validation | Tune hyperparameters | 15-20% |
| Test | Final performance check, used once | 15-20% |


```python
from sklearn.model_selection import train_test_split
train_df, temp_df = train_test_split(df_raw, test_size=0.30, random_state=42)
val_df, test_df = train_test_split(temp_df, test_size=0.50, random_state=42)
print(len(train_df), len(val_df), len(test_df))
```

## 8. Preprocessing Pipeline

1. Load raw data
2. Fix data types
3. Standardize categorical values
4. Remove duplicate records
5. Handle outliers and invalid values
6. Handle missing values
7. Drop irrelevant features
8. Split into train, validation, test
9. Fit transformations on training data only
10. Encode features for modeling
